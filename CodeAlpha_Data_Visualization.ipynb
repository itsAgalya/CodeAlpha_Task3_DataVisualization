```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("CodeAlpha_Task2_EDA.csv")

print(df.head())
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="Rating", data=df, order=["One", "Two", "Three", "Four", "Five"])

plt.title("Book Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.hist(df["Price"], bins=5)

plt.title("Book Price Distribution")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.show()
```

```python
plt.figure(figsize=(8, 5))
rating_order = ["One", "Two", "Three", "Four", "Five"]
sns.boxplot(
    x="Rating",
    y="Price",
    data=df,
    order=rating_order
)
plt.title("Book Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.show()
```

```python
rating_order = ["One", "Two", "Three", "Four", "Five"]
avg_price = df.groupby("Rating")["Price"].mean().reindex(rating_order)
plt.figure(figsize=(8, 5))
avg_price.plot(kind="bar")
plt.title("Average Book Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)
plt.show()
```

```python
print("Data Visualization Summary")
print("----------------------------")
print("Total Books:", len(df))
print("Average Price: £", round(df["Price"].mean(), 2))
print("Minimum Price: £", round(df["Price"].min(), 2))
print("Maximum Price: £", round(df["Price"].max(), 2))
print("Most Common Rating:", df["Rating"].value_counts().idxmax())
```
